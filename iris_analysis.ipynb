```python
import os

os.makedirs('output', exist_ok=True)

species_summary = df.groupby('species').mean()
species_summary.to_csv('output/species_summary.csv')

filtered_df = df[df['sepal_length'] > 6]
filtered_df.to_csv('output/filtered_iris.csv', index=False)

os.listdir('output')
```

> - Created the `output/` directory for saving generated CSV reports.
> - Exported species-level summary statistics to `output/species_summary.csv`.

> - Verified successful creation and structure of both output files.
